# Merge

```python
import pandas as pd
var1 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
var2 = pd.DataFrame({"A":[1,2,3,5], "C":[21,22,23,24]})

pd.merge(var1,var2, on="A")
```

```python
var3 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
var4 = pd.DataFrame({"A":[1,2,3,5], "C":[21,22,23,24]})

# pd.merge(var3,var4, how= "inner")
# pd.merge(var3,var4, how= "outer")
# pd.merge(var3,var4, how= "left")
# pd.merge(var3,var4, how= "right")
# pd.merge(var3,var4, how= "outer", indicator=True)
pd.merge(var3,var4,left_index=True,right_index=True,suffixes=("Name","ID"))
```

```python
var3 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
var4 = pd.DataFrame({"A":[1,2,3,5], "B":[21,22,23,24]})

pd.merge(var3,var4,left_index=True,right_index=True,suffixes=("Name","ID"))
```

# Concat

```python
sr1=pd.Series([1,2,3,4])
sr2=pd.Series([21,31,41,51])
pd.concat([sr1,sr2])
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
d2 = pd.DataFrame({"A":[1,2,3,5], "B":[21,22,23,24]})

# pd.concat([d1,d2])
pd.concat([d1,d2],axis=1) #According to column merge
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
d2 = pd.DataFrame({"A":[1,2], "B":[21,22]})
# pd.concat([d1,d2],axis=1,join="outer")
pd.concat([d1,d2],axis=1,join="inner")
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4], "B":[11,12,13,14]})
d2 = pd.DataFrame({"A":[1,2,3,4], "B":[21,22,23,24]})
# pd.concat([d1,d2],axis=1,join="outer")
pd.concat([d1,d2],axis=0,keys=["d1","d2"])
```

```python
d1 = pd.DataFrame({"D":[1,2,3,4]})
d2 = pd.DataFrame({"A":[1,2,3,4], "B":[21,22,23,24]})

pd.concat([d1,d2])
```
