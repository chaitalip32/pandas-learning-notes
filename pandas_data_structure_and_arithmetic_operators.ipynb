## 3. Pandas data structures

Pandas mainly provides two data structures:

- **Series:** one-dimensional labelled data, like a single column.
- **DataFrame:** two-dimensional labelled data, arranged in rows and columns like a spreadsheet or database table.

A label used to identify a row is called an **index**. By default, Pandas usually creates an index starting at `0`.

### 3.1 Creating a Series from a list

A Series stores values along with an index. If you do not provide an index, Pandas creates `0, 1, 2, ...` automatically.

```python
s = pd.Series([12, 20, 30, 40])
print(s)
print(type(s))
```

**Remember:** the left side of the output is the index; the right side contains the values. The final line is the data type of the whole object, `pandas.Series`.

### 3.2 Creating a Series with a custom index

Use the `index=` argument to assign meaningful labels. The number of index labels must match the number of values.

```python
data = [3, 5, 7, 9]
labels = ["one", "two", "three", "four"]

s = pd.Series(data, index=labels, name="Python data", dtype=float)
print(s)
```

- `data` supplies the values.
- `index=labels` supplies the labels.
- `name=` gives the Series a descriptive name.
- `dtype=float` asks Pandas to store the values as floating-point numbers, so values appear like `3.0`.

Use named arguments such as `index=` because they make your code easier to read.

### 3.3 Creating a Series with one repeated value

When a scalar value is supplied with an index, Pandas repeats that value for each index label.

```python
s = pd.Series(12, index=[1, 2, 3, 4])
print(s)
print(type(s))
```

### 3.4 Creating a Series from a dictionary

Dictionary keys become the Series index, and dictionary values become the Series values.

```python
scores = {"Amit": 85, "Priya": 92, "Rahul": 78}
s = pd.Series(scores)

print(s)
print(s["Priya"])  # Select the value labelled Priya
```

**Note about your earlier example:** a dictionary whose values are themselves dictionaries or lists can produce a Series containing objects rather than a simple numeric Series. For learning basic Series creation, start with a dictionary mapping each label to one value, as above.

### 3.5 Selecting Series values

Use square brackets with a label or position. For explicit label-based selection, `.loc[]` is clear; for position-based selection, `.iloc[]` is clear.

```python
s = pd.Series([10, 20, 30], index=["a", "b", "c"])

print(s.loc["b"])   # By label: 20
print(s.iloc[1])    # By position: 20
print(s.iloc[0:2])  # Positions 0 and 1
```

### 3.6 Series arithmetic and index alignment

Pandas arithmetic matches Series values by their **index labels**, not simply by their position. Labels present in only one Series produce a missing value (`NaN`) in the result.

```python
s1 = pd.Series(12, index=[1, 2, 3, 4, 5, 6, 7])
s2 = pd.Series(12, index=[1, 2, 3, 4])

print(s1 + s2)
```

In this example, indexes `1` through `4` exist in both Series, so their values add to `24`. Indexes `5`, `6`, and `7` exist only in `s1`, so the result is `NaN` for those labels.

`NaN` means a value is missing or unavailable. You can fill missing results with a chosen value when that is appropriate, for example `(s1 + s2).fillna(0)`. Do not replace missing values with zero automatically unless zero is meaningful for your task.

## 4. DataFrames

A **DataFrame** is a two-dimensional table with rows and columns. Each column can have its own data type. It is commonly used to work with real-world datasets.

### 4.1 Creating a DataFrame from a dictionary

Dictionary keys become column names. Each list supplies the values for its column. Lists should normally have the same length.

```python
data = {
    "Name": ["Amit", "Priya", "Rahul"],
    "Age": [22, 24, 21],
    "City": ["Pune", "Mumbai", "Nashik"]
}

df = pd.DataFrame(data)
print(df)
```

Explanation:
- `pd.DataFrame(data)` creates the table.
- `"Name"`, `"Age"`, and `"City"` become column names.
- Each list provides values for its column.
- Pandas creates a default row index (`0`, `1`, `2`).

To select a column, use `df["Name"]`. To select a particular value by row position and column label, use `df.loc[1, "Name"]` or `df.iloc[1, 0]`.

```python
print(df["Name"])           # One column: returns a Series
print(df.loc[1, "Name"])    # Row label 1, column Name
print(df.iloc[1, 0])        # Second row, first column
print(type(df))             # pandas.DataFrame
print(type(data))           # dict: the original Python object
```

### 4.2 Creating a DataFrame from a list of lists

Each inner list becomes one row. If you do not provide column names, Pandas assigns numeric column labels.

```python
rows = [[1, 2, 3, 4], [2, 3, 4, 5]]
df_from_list = pd.DataFrame(rows, columns=["A", "B", "C", "D"])
print(df_from_list)
```

### 4.3 Creating a DataFrame from Series

When Series are used as columns, Pandas aligns their values by index labels. If indexes do not match, missing values may appear.

```python
columns = {
    "a": pd.Series([1, 2, 3]),
    "b": pd.Series([4, 5, 6])
}
df_from_series = pd.DataFrame(columns)
print(df_from_series)
```

## 5. Inspecting a DataFrame

These are useful first checks when you receive a dataset.

```python
print(df.head())       # First five rows (or fewer if the table is smaller)
print(df.shape)        # (number_of_rows, number_of_columns)
print(df.columns)      # Column labels
print(df.index)        # Row labels
print(df.dtypes)       # Data type of each column
print(df.info())       # Compact summary; prints information and returns None
print(df.describe())   # Summary statistics for numeric columns
```

## 6. Selecting rows and columns

- `df["Age"]` selects one column as a Series.
- `df[["Name", "City"]]` selects multiple columns as a DataFrame.
- `.loc[]` selects using labels.
- `.iloc[]` selects using integer positions.

```python
print(df["Age"])
print(df[["Name", "City"]])
print(df.loc[0])              # First row by its label
print(df.loc[0:1, ["Name", "Age"]])  # Labels 0 through 1, inclusive
print(df.iloc[0:2, 0:2])      # First two rows and first two columns
```

## 7. Filtering with logical conditions

A comparison such as `df["Age"] >= 22` creates a Series of Boolean values (`True` or `False`). Use that Boolean Series to filter rows.

Common comparison operators:
- `==` equal to
- `!=` not equal to
- `>` greater than
- `<` less than
- `>=` greater than or equal to
- `<=` less than or equal to

For multiple conditions, use `&` for AND and `|` for OR. Put each condition inside parentheses. Use `~` to negate a Boolean condition.

```python
df["Age is at least 22"] = df["Age"] >= 22
print(df)

adults_22_plus = df[df["Age"] >= 22]
print(adults_22_plus)

pune_or_older = df[(df["City"] == "Pune") | (df["Age"] >= 24)]
print(pune_or_older)
```

**Important:** use `&` and `|` for element-by-element conditions in Pandas, not Python's `and` and `or`. Parentheses are important because of operator precedence. If you want to keep the original DataFrame unchanged, create a filtered result as shown instead of adding a new column.

## 8. Arithmetic operations on DataFrame columns

You can create new columns by applying arithmetic operators to existing columns.

```python
numbers = pd.DataFrame({
    "A": [10, 20, 30, 40],
    "B": [15, 16, 17, 18]
})

numbers["Add"] = numbers["A"] + numbers["B"]
numbers["Sub"] = numbers["A"] - numbers["B"]
numbers["Multiply"] = numbers["A"] * numbers["B"]
numbers["Divide"] = numbers["A"] / numbers["B"]

print(numbers)
```

The operation is performed row by row because the two columns share the same row index. For example, the first `Add` value is `10 + 15 = 25`.

Logical comparisons also work on columns, for example `numbers["A"] >= 20`, and return Boolean values.

## 9. Missing values

Missing data is common in real datasets. Pandas commonly displays missing numeric results as `NaN`.

Useful methods:
- `df.isna()` identifies missing values.
- `df.isna().sum()` counts missing values in each column.
- `df.dropna()` returns rows containing no missing values.
- `df.fillna(value)` returns data with missing values filled using `value`.

Choose how to handle missing data based on what it means in the dataset.

```python
sample = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Score": [85, None, 78]
})

print(sample.isna())
print(sample.isna().sum())
print(sample.fillna({"Score": sample["Score"].mean()}))
```
